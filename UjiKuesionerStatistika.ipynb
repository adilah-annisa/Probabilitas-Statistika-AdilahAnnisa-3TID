### **A. CODINGAN PRAKTIKUM MENGGUNAKAN KUESIONER GOOGLE FORM**

```python
# Langkah 1 (Import Library)
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
# Langkah 2 (Upload File Excel)
up = files.upload()
nama_file = list(up.keys()) [0]
print("File terbaca:", nama_file)
```

```python
# Langkah 3 (Membaca Data dan Memisahkan Butir)
JUMLAH_KOLOM_DEMOGRAFI = 5
df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]

# Buang kolom yang bukan butir Likert (misalnya Durasi)
butir = butir.loc[:, butir.columns.str.contains(r" \[")]

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", butir.isna().sum().sum())
```

```python
# Langkah 4 (Konversi Likert menjadi Angka)
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}

num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

# Harus 0. Jika lebih dari 0, ada jawaban yang tidak terbaca.
print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
# Langkah 5 (Mengelompokkan Buktir per Dimensi)
dimensi = {}
for kolom in butir.columns:
  nama = re.match(r"^(.*?) \[", kolom).group(1)
  dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
  print(f"{nama}: {len(daftar)} butir")
```

```python
# Langka 6 (Uji Validitas (Korelasi Pearson Butir-Total))
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
  x = num [daftar]
  total = x.sum(axis=1)
  for i, kolom in enumerate(daftar, 1):
    r, p = stats.pearsonr(x[kolom], total)
    r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
    hasil_validitas.append({
        "Dimensi": nama,
        "Butir": f"B{i}",
        "r hitung": round(r, 3),
        "Corrected r": round(r_koreksi, 3),
        "p-value": round(p,4),
        "Keterangan": "Valid" if r > r_tabel else "Tidak Valid",
    })

  tabel_validitas = pd.DataFrame(hasil_validitas)
  tabel_validitas
```

```python
# Langkah 7 (Uji Reliabilitas (Cronbach's Alpha))
def cronbach_alpha(x):
  k = x.shape[1]
  return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
  return("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
         else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
  x = num[daftar]
  a = cronbach_alpha(x)
  hasil_reliabilitas.append({
      "Dimensi": nama,
      "Jumlah butir": len(daftar),
      "Cronbach's alpha": round(a,3),
      "Kategori": kategori(a),
  })

  for i, kolom in enumerate(daftar, 1):
    if len(daftar) > 2:
      alpha_dihapus.append({
          "Dimensi": nama,
          "Butir": f"B{i}",
          "Cronbach's alpha": round(cronbach_alpha(x.drop(columns=kolom)), 3)})

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
# Langkah 8 (Menggabungkan dan Mengunduh Hasil)
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"],
how = "left")

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
  gabung.to_excel(w, sheet_name="Validitas", index=False)
  tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```

### **B. CODINGAN LATIHAN UJI VALIDITAS DAN REABILITAS**

```python
# Latihan 1: butir tidak valid, r hitung, dan r tabel
tv = tabel_validitas.copy()
tv["r tabel"] = round(r_tabel, 3)

tidak_valid = tv[tv["Keterangan"] == "Tidak valid"]
print("Butir tidak valid (r hitung <= r tabel):", len(tidak_valid))

# 4 butir dengan r hitung terendah (paling dekat dengan batas)
display(tv.sort_values("r hitung").head(4)[
    ["Dimensi", "Butir", "r hitung", "r tabel", "Corrected r"]])
```

```python
# Latihan 2: r hitung vs corrected r
total = len(tabel_validitas)
valid_r = int((tabel_validitas["r hitung"] > r_tabel).sum())
valid_c = int((tabel_validitas["Corrected r"] >= 0.30).sum())

display(pd.DataFrame({
    "Kriteria": [f"r hitung > r tabel ({r_tabel:.3f})", "Corrected r ≥ 0,30"],
    "Butir valid": [valid_r, valid_c],
    "Butir tidak valid": [total - valid_r, total - valid_c],
}))

# Butir yang lolos kriteria pertama tapi gugur di kriteria kedua
display(tabel_validitas[tabel_validitas["Corrected r"] < 0.30][
    ["Dimensi", "Butir", "r hitung", "Corrected r"]])
```

```python
# Latihan 3: buang Dimensi 2 B6, hitung ulang
nama_dim = list(dimensi.keys())[1]          # Aspek Psikologis
kol = dimensi[nama_dim]
kol_baru = [k for k in kol if k != kol[5]]  # buang B6

sebelum = cronbach_alpha(num[kol])
sesudah = cronbach_alpha(num[kol_baru])

display(pd.DataFrame({
    "Kondisi": ["Sebelum (6 butir)", "Sesudah (5 butir)"],
    "Cronbach's alpha": [round(sebelum, 3), round(sesudah, 3)],
    "Kategori": [kategori(sebelum), kategori(sesudah)],
}))

# Uji ulang validitas butir yang tersisa
x = num[kol_baru]
total = x.sum(axis=1)
n2 = len(x)
t2 = stats.t.ppf(0.975, n2 - 2)
rt2 = t2 / np.sqrt((n2 - 2) + t2**2)

hasil = []
for i, k in enumerate(kol_baru, 1):
    r, _ = stats.pearsonr(x[k], total)
    rc, _ = stats.pearsonr(x[k], total - x[k])
    hasil.append({"Butir": f"B{i}", "r hitung": round(r, 3),
                  "Corrected r": round(rc, 3),
                  "Keterangan": "Valid" if r > rt2 else "Tidak valid"})
display(pd.DataFrame(hasil))
```

```python
# Latihan 4A: ubah n dengan subset acak
def validitas(data):
    n = len(data)
    t = stats.t.ppf(0.975, n - 2)
    rt = t / np.sqrt((n - 2) + t**2)
    jml_r = jml_c = 0
    for nama, daftar in dimensi.items():
        x = data[daftar]
        total = x.sum(axis=1)
        for k in daftar:
            r, _ = stats.pearsonr(x[k], total)
            rc, _ = stats.pearsonr(x[k], total - x[k])
            jml_r += r > rt
            jml_c += rc >= 0.30
    return rt, jml_r, jml_c

hasil = []
for m in [38, 30, 20, 15, 10]:
    sub = num if m == 38 else num.sample(m, random_state=1)
    rt, a, b = validitas(sub)
    hasil.append({"n": m, "df": m - 2, "r tabel": round(rt, 3),
                  "Valid (r hitung)": int(a),
                  "Valid (corrected r ≥ 0,30)": int(b)})
display(pd.DataFrame(hasil))
```

```python
# Latihan 4B: 100 kali undian acak per ukuran sampel
import warnings
warnings.filterwarnings("ignore")

rata = []
for m in [30, 20, 15, 10]:
    jml = [int(validitas(num.sample(m, random_state=s))[1]) for s in range(1, 101)]
    rata.append({"n": m, "Rata-rata butir valid": round(np.mean(jml), 1),
                 "Minimum": min(jml), "Maksimum": max(jml)})
display(pd.DataFrame(rata))

konstan = {}
for m in [30, 20, 15, 10]:
    hitung = 0
    for s in range(1, 101):
        sub = num.sample(m, random_state=s)
        hitung += int((sub.nunique() == 1).sum())   # jumlah butir konstan
    konstan[m] = hitung
pd.DataFrame({"n": konstan.keys(), "Total butir konstan (100 undian)": konstan.values()})
```
